```python
import matplotlib.pyplot as plt
import numpy as np
import time
import sys
```

## zad.1

```python
def pade_1(x):
    return (6 - 2 * x) / (6 + 4 * x + x ** 2)

def pade_2(x):
    return (6 - 4 * x + x ** 2) / (6 + 2 * x)

def exp_minus(x):
    return np.e**(-x)

def error(x, function):
    return abs(function(x) - exp_minus(x))

x = np.arange(-5.1,12.9,0.2)

#1
plt.plot(x, pade_1(x), color = 'blue', label = 'Pade_1')
plt.plot(x, pade_2(x), color = 'orange', label = 'Pade_2')
plt.plot(x, exp_minus(x), color = 'red', linestyle = '--', label = 'e^(-1)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.title('Comparison between two Pade approximations')
plt.grid()

plt.savefig('comparison.png')
plt.show()

#2
window, plots = plt.subplots(nrows = 1, ncols = 2)

plots[0].plot(x, pade_1(x), color = 'blue', label = 'Pade_1')
plots[0].set_xlabel('x')
plots[0].set_ylabel('y')
plots[0].legend()
plots[0].set_title('Pade_1 plot')
plots[0].grid()

plots[1].plot(x, pade_2(x), color = 'orange', label = 'Pade_2')
plots[1].set_xlabel('x')
plots[1].set_ylabel('y')
plots[1].legend()
plots[1].set_title('Pade_2 plot')
plots[1].grid()

plt.tight_layout()
plt.savefig('pade_plots.png')
plt.show()

#3
window_1, plots_1 = plt.subplots(nrows = 1, ncols = 2)

plots_1[0].plot(x, error(x,pade_1), color = 'blue',  label = 'Pade_1 error')
plots_1[0].set_xlabel('x')
plots_1[0].set_ylabel('y')
plots_1[0].legend()
plots_1[0].set_title('Pade_1 error plot')
plots_1[0].grid()


plots_1[1].plot(x, error(x,pade_2), color = 'orange', label = 'Pade_2 error')
plots_1[1].set_xlabel('x')
plots_1[1].set_ylabel('y')
plots_1[1].legend()
plots_1[1].set_title('Pade_1 error plot')
plots_1[1].grid()

plt.tight_layout()
plt.savefig('pade_errors.png')
plt.show()

```

## zad.2

```python
def machine_epsilon():
    eps = 1
    while 1 + eps != 1:
        eps = eps/2
        
    return eps * 2

print(f"{machine_epsilon()}")
print(f"{sys.float_info.epsilon}")
```

```python
def er(x):
    return x/100 * 100 - x

print(er(7))

for i in range(0,51):
    e = er(i)
    if e != 0:
        print(f'{i} ---> {e}')

print(er(36))
```

## zad.4

```python
def IEEE_to_normal(bits):
    sign = int(bits[0])
    E_127 = int(bits[1:9],2) - 127
    M = int(bits[9:],2) / (2**(23))
    return (-1) ** (sign) * (1+M) * 2 ** (E_127)

IEE = '00111111110110011001100110011001'
print(IEEE_to_normal(IEE))

abs_error = abs(1.7 - IEEE_to_normal(IEE))
relative_error = abs_error / 1.7

print(abs_error)
print(relative_error)
```

## zad.6

```python
def horner_polynomial(x):
    return x * ( x * ( x * ( x * (6) + 5) - 13) + 1 ) + 1

x = np.arange(-10,10,0.0001)
start = time.perf_counter()
y = horner_polynomial(x)
stop = time.perf_counter()

total_time = stop - start
print(total_time)
```
