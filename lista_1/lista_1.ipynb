```python
import matplotlib.pyplot as plt
import numpy as np
import time
```

## zad.1

```python
def pade_1(x):
    return (6 - 2 * x) / (6 + 4 * x + x ** 2)

def pade_2(x):
    return (6 - 4 * x + x ** 2) / (6 + 2 * x)

def exp_minus(x):
    return np.e**(-x)

def error(x, function):
    return abs(function(x) - exp_minus(x))

x = np.arange(-5.1,12.9,0.2)

#1
plt.plot(x, pade_1(x), color = 'blue', label = 'Pade_1')
plt.plot(x, pade_2(x), color = 'orange', label = 'Pade_2')
plt.plot(x, exp_minus(x), color = 'red', linestyle = '--', label = 'e^(-1)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.title('Comparison between two Pade approximations')
plt.grid()

plt.savefig('comparison.png')
plt.show()

#2
window, plots = plt.subplots(nrows = 1, ncols = 2)

plots[0].plot(x, pade_1(x), color = 'blue', label = 'Pade_1')
plots[0].set_xlabel('x')
plots[0].set_ylabel('y')
plots[0].legend()
plots[0].set_title('Pade_1 plot')
plots[0].grid()

plots[1].plot(x, pade_2(x), color = 'orange', label = 'Pade_2')
plots[1].set_xlabel('x')
plots[1].set_ylabel('y')
plots[1].legend()
plots[1].set_title('Pade_2 plot')
plots[1].grid()

plt.tight_layout()
plt.savefig('pade_plots.png')
plt.show()

#3
window_1, plots_1 = plt.subplots(nrows = 1, ncols = 2)

plots_1[0].plot(x, error(x,pade_1), color = 'blue',  label = 'Pade_1 error')
plots_1[0].set_xlabel('x')
plots_1[0].set_ylabel('y')
plots_1[0].legend()
plots_1[0].set_title('Pade_1 error plot')
plots_1[0].grid()


plots_1[1].plot(x, error(x,pade_2), color = 'orange', label = 'Pade_2 error')
plots_1[1].set_xlabel('x')
plots_1[1].set_ylabel('y')
plots_1[1].legend()
plots_1[1].set_title('Pade_1 error plot')
plots_1[1].grid()

plt.tight_layout()
plt.savefig('pade_errors.png')
plt.show()

```
